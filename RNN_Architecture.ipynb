```python
import tensorflow
from tensorflow import keras
from keras import Sequential
from keras.layers import Dense,SimpleRNN
```

```python
model=Sequential([
    SimpleRNN(3,input_shape=(4,5)),
    Dense(1,activation='sigmoid')
])
```

```python
model.summary()
```

```python
print(model.get_weights()[0].shape)
model.get_weights()[0]
```
